## A2. Explain vectorization. Why does replacing an explicit Python loop with a vectorized NumPy operation matter with large datasets? [2 marks]

## Answer:
Vectorization means performing an operation on the whole NumPy array at once instead of using a Python loop for each value. It is useful because NumPy can perform these operations faster and more efficiently. When the dataset is large, vectorization reduces the time needed for calculations and makes the program more efficient.



## A3. For compatible 2D arrays A and B, explain A * B vs A @ B. Describe one neural-network situation for each. [2 marks]

## Answer:
A * B performs element-wise multiplication, which means each element of A is multiplied by the corresponding element of B. A @ B performs matrix multiplication, where rows and columns are multiplied to produce a new matrix. In a neural network, * can be used for element-wise operations such as applying a mask, while @ is commonly used to multiply input data with a weight matrix.


## A4. What is broadcasting in NumPy? State the general rule used to decide whether arrays of different shapes can be combined element-wise. [2 marks]

## Answer:
Broadcasting is a NumPy feature that allows arrays with different shapes to be used together in element-wise operations. NumPy compares their shapes starting from the last dimension. The dimensions are compatible when they are equal or when one of them is 1. If the dimensions do not satisfy this rule, NumPy gives a shape error.

## A5. Explain flatten() vs ravel(). In a memory-constrained application processing large images, which would you prefer and why? [2 marks]

## Answer:
flatten() converts an array into a 1D array and always creates a new copy of the data. ravel() also converts an array into 1D, but it usually returns a view of the original array instead of making a copy. For a memory-constrained application working with large images, I would prefer ravel() because it usually uses less extra memory.

For a 2D array, axis=0 means the operation is performed column-wise, while axis=1 means it is performed row-wise

## C2. Write a function standardize(x) that takes a 1D NumPy array and returns its z-score standardized version using only NumPy. 

````python
import numpy as np
def standardize(x):
    mean=np.mean(x)
    standard_dev=np.std(x)
    return (x-mean)/standard_dev
x=np.array([25,35,10,45,40])
res=standardize(x)
print(res)
````

## Explanation: 
The np.mean() function calculates the mean and np.std() calculates the standard deviation of the array and converts each value into its z-score using (x - mean) / std.

## C5. Implement y_pred = X @ w + b for a feature matrix X, weight vector w and bias b. Then compute MSE using y_true

````python
import numpy as np

X = np.array([[1, 2],
              [3, 4],
              [5, 6]])

w = np.array([0.5, 1])
b = 2

y_true = np.array([4, 8, 10])

y_pred = X @ w + b

mse = np.mean((y_true - y_pred) ** 2)

print("Predicted values:", y_pred)
print("MSE:", mse)
````

## Explanation:
X @ w performs matrix multiplication between the feature matrix and weight vector. Then b is added to get the predicted values. Finally, MSE is calculated by finding the mean of the squared differences between the actual values y_true and predicted values y_pred.

## C7. Given a grayscale image as a 2D NumPy array with pixel values in [0, 255], normalize the values to [0, 1] using min-max normalization.

````python
import numpy as np

image = np.array([[0, 50, 100],
                  [150, 200, 255]])

normalized = (image - image.min()) / (image.max() - image.min())

print(normalized)
````

## Explanation:

## D1. Data Preprocessing Pipeline

You are given a raw 1D or 2D NumPy array containing NaN values. Describe, step by step, the NumPy operations you would apply to:

(a) detect missing values,
(b) handle them appropriately and justify your strategy, and
(c) normalize or standardize the clean data before it is fed to a model.

Implement the full pipeline in code on a sample array of your own choosing, including at least one NaN.

## Answer:

- First, I would detect the missing values using `np.isnan()`.
- Then, I would handle the missing values by replacing each NaN with the mean of its column. This is useful because we can keep the data instead of removing the complete row.
- After handling the missing values, I would standardize the data using the mean and standard deviation of each column.

````python
data = np.array([[10, 20],
                 [15, np.nan],
                 [30, 40]]) 

# (a) Detect missing values
missing = np.isnan(data)
print("Missing values:")
print(missing) 
````

````python
# (b) Handle missing values using column mean
column_mean = np.nanmean(data, axis=0)
rows, cols = np.where(np.isnan(data))
data[rows, cols] = column_mean[cols]
print("After handling NaN:")
print(data)
````

````python
# (c) Standardize the clean data
mean = np.mean(data, axis=0)
std = np.std(data, axis=0)
standardized_data = (data - mean) / std
print("Standardized data:")
print(standardized_data)
````

## E1. Per-Column Min–Max Normalization [5 marks]
Implement min-max normalization for a 2D feature matrix on a per-column (per-feature) basis, rather than using a single global min/max for the whole array. Your solution must be fully vectorized, with no explicit loops over rows or columns.

````python
X = np.array([[10, 100],
              [20, 200],
              [30, 300],
              [40, 400]])
min_values = np.min(X, axis=0)
max_values = np.max(X, axis=0)
X_normalized = (X - min_values) / (max_values - min_values)
print(X_normalized)
````

- Here, axis=0 calculates the minimum and maximum for each column separately.
- The operation is fully vectorized, so there are no explicit loops.

## E2. Three-Level Score Classification [5 marks]
Given a 1D array of exam scores, use a single vectorized expression (no if/else statements and no explicit loops) to classify each score as:
Fail (< 40), Average (40–70 inclusive), or Excellent (> 70).
Hint: You may need to nest or combine np.where() calls.

````python
scores = np.array([25, 40, 55, 70, 85, 95])
result = np.where(scores < 40, "Fail", np.where(scores <= 70, "Average", "Excellent"))
print(result)
````
